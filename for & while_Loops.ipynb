## 1. Why do we need loops?

Imagine printing "Hello" 5 times. Without loops:
```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```
This is repetitive. **Loops** let us repeat a block of code without rewriting it.

Python has two main loops:
- **`for` loop** → used when we know how many times to repeat, or when we want to go through items one by one.
- **`while` loop** → used when we repeat *until a condition becomes False* (number of repetitions may not be known in advance).

````python
# Without loop (repetitive)
print("Hello")
print("Hello")
print("Hello")
````

````python
# With loop (clean)
for i in range(3):
    print("Hello")
````

## 2. `for` loop basics — using `range()`

`range(start, stop, step)` generates numbers. `stop` is **not included**.

- `range(5)` → 0,1,2,3,4
- `range(2,6)` → 2,3,4,5
- `range(0,10,2)` → 0,2,4,6,8

````python
# basic for loop with range
for i in range(5):
    print("Count:", i)
````

````python
# for loop with start, stop
for i in range(2, 6):
    print(i)
````

````python
# for loop with step (using arithmetic operator too)
for i in range(0, 10, 2):
    print(i, "-> square is", i * i)   # using ** or * operator
````

````python
# for loop + conditional statement -> print even/odd
for i in range(1, 11):
    if i % 2 == 0:          # modulus operator
        print(i, "is Even")
    else:
        print(i, "is Odd")
````

## 3. `for` loop with data structures

You can loop directly over a **list, tuple, string, or dictionary** — no need for `range()`.

````python
# for loop over a list
fruits = ["apple", "banana", "mango", "grapes"]
for fruit in fruits:
    print("I like", fruit)
````

````python
# for loop over a string (string is a sequence of characters)
name = "Python"
for ch in name:
    print(ch)
````

````python
# Practical: count vowels in a string using for loop + conditional + list
word = "Programming"
vowels = "aeiouAEIOU"
count = 0

for ch in word:
    if ch in vowels:
        count += 1          # same as count = count + 1

print("Number of vowels in", word, "=", count)
````

````python
# for loop over a dictionary
marks = {"Math": 90, "Science": 85, "English": 78}

for subject in marks:
    print(subject, "->", marks[subject])

# or using .items()
for subject, score in marks.items():
    print(subject, "scored", score)
````

## 4. `while` loop basics

`while condition:` keeps running the block **as long as the condition is True**.

⚠️ Make sure the condition eventually becomes False, otherwise it becomes an **infinite loop**.

````python
# basic while loop - countdown
n = 5
while n > 0:
    print(n)
    n -= 1          # important: update the variable, else infinite loop
print("Liftoff!")
````

````python
# while loop with input - keep asking until user enters 'stop'
# (Run this cell and type a few words, then type 'stop')
word = ""
while word.lower() != "stop":
    word = input("Enter a word (type 'stop' to end): ")
    if word.lower() != "stop":
        print("You typed:", word)
````

````python
# Practical: sum of numbers entered by user until they enter 0
total = 0
num = int(input("Enter a number (0 to stop): "))

while num != 0:
    total += num
    num = int(input("Enter a number (0 to stop): "))

print("Total sum =", total)
````

## 5. `break` and `continue`

- **`break`** → stops the loop immediately.
- **`continue`** → skips the current iteration and moves to the next one.

````python
# break example - stop loop when we find the number
numbers = [4, 8, 15, 16, 23, 42]
search = 16

for n in numbers:
    if n == search:
        print("Found", search, "in the list!")
        break
    print("Checked", n, "- not a match")
````

````python
# continue example - skip printing negative numbers
numbers = [10, -5, 20, -3, 8]

for n in numbers:
    if n < 0:
        continue        # skip rest of code for this value
    print(n)
````

## 6. Nested loops (loop inside a loop)

Useful for grids, tables, and patterns.

````python
# multiplication table using nested for loops
for i in range(1, 4):
    for j in range(1, 4):
        print(i, "x", j, "=", i * j)
    print("---")
````

````python
# Number pyramid — each row prints DIFFERENT numbers 
for i in range(1, 5):
    for j in range(1, i+1):
        print(j, end=' ')
    print()
````

````python
#  Hollow square — border is '*', inside is blank (needs per-position decision)
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print('*', end=' ')
        else:
            print(' ', end=' ')
    print()
````

## 7. Practical Mini-Programs

Combining loops with input, conditionals, operators and data structures.

````python
# Practical 1: Simple calculator that keeps running until user chooses to exit
while True:
    print("\n1. Add  2. Subtract  3. Multiply  4. Divide  5. Exit")
    choice = input("Choose an option (1-5): ")

    if choice == "5":
        print("Exiting calculator. Bye!")
        break

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == "1":
        print("Result:", a + b)
    elif choice == "2":
        print("Result:", a - b)
    elif choice == "3":
        print("Result:", a * b)
    elif choice == "4":
        if b != 0:
            print("Result:", a / b)
        else:
            print("Error: Division by zero!")
    else:
        print("Invalid choice, try again.")
````

````python
# Practical 2: Check if a number is prime (loops + conditionals + operators)
num = int(input("Enter a number: "))
is_prime = True

if num < 2:
    is_prime = False
else:
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

if is_prime:
    print(num, "is a Prime number")
else:
    print(num, "is NOT a Prime number")
````

````python
# Practical 3: Grade report card using dictionary + for loop + conditionals
students = {"Anu": 92, "Ravi": 67, "Meera": 45, "Kiran": 78}

for name, score in students.items():
    if score >= 90:
        grade = "A"
    elif score >= 75:
        grade = "B"
    elif score >= 50:
        grade = "C"
    else:
        grade = "Fail"
    print(name, "-> Score:", score, "-> Grade:", grade)
````

````python
# Practical 4: Reverse a list manually using a while loop (without using [::-1])
numbers = [1, 2, 3, 4, 5]
reversed_list = []

i = len(numbers) - 1
while i >= 0:
    reversed_list.append(numbers[i])
    i -= 1

print("Original:", numbers)
print("Reversed:", reversed_list)
````
