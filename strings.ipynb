# Strings in Python — Practical Notebook

## Creating Strings

```
single = 'Python'
double = "It's a great day!"
multi = """Python
Programming
Language"""

print(single)
print(double)
print(multi)
```

## Indexing

```
text = "Python"
print(text[0])
print(text[2])
print(text[-1])
print(text[-3])
```

## Slicing

```
text = "Python"
print(text[0:3])
print(text[0:2])
print(text[:3])
print(text[2:])
print(text[:])
print(text[-3:])
```

```
text = "Python"
print(text[::2])     # every 2nd character
print(text[::-1])    # reverse the string
```

## Concatenation & Repetition

```
first = "Hello"
second = "World"
print(first + " " + second)

print("Python " * 3)
```

## Membership Operator

```
text = "Python"
print("P" in text)
print("Java" in text)
```

## Common String Methods

```
text = "python"
print(text.upper())
```

```
text = "PYTHON"
print(text.lower())
```

```
text = "python programming"
print(text.title())
```

```
text = "python programming"
print(text.capitalize())
```

```
text = "   Python   "
print(text.strip())
```

```
text = "I like Java"
print(text.replace("Java", "Python"))
```

```
text = "Python Java SQL"
print(text.split())
```

```
languages = ["Python", "Java", "SQL"]
print("-".join(languages))
```

```
text = "Python"
print(text.find("t"))
```

```
text = "banana"
print(text.count("a"))
```

```
text = "Python"
print(text.startswith("Py"))
print(text.endswith("on"))
```

## String Formatting

```
name = "John"
age = 25
print(f"My name is {name} and I am {age} years old")
```

```
name = "John"
age = 25
print("My name is {} and I am {} years old".format(name, age))
```

## Escape Characters

```
print("Python\nProgramming")
print("Tab\tSpace")
print('It\'s fine')
print("She said \"hi\"")
print("C:\\Users\\path")
```
