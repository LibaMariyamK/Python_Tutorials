## List Methods

````python
fruits = ["Apple", "Banana", "Mango"]
new = "kiwi"
fruits.insert(1,new)
fruits
````

````python
my_list = [1, 2]
my_list.append(3)
print(my_list)
````

````python
my_list = [1, 2]
my_list.extend([3, 4])
print(my_list)
````

````python
my_list = [1, 3]
my_list.insert(1, 2)
print(my_list)
````

````python
my_list = [10, 22, 22, 40, 34, 33, 56]
a = 40
my_list.remove(a)
print(my_list)
````

````python
my_list = [1, 2, 3]
removed = my_list.pop(0)

print(my_list)
removed
````

````python
my_list = [1, 2, 3]
my_list.clear()
print(my_list)
````

````python
my_list = [3, 1, 2]
my_list.sort()
print(my_list)
````

````python
my_list = [3, 1, 2]
my_list.sort(reverse=True)
print(my_list)
````

````python
my_list = [10, 33, 4,22, 56,6]
my_list.reverse() # my_list[::-1]
print(my_list)
````

````python
my_list = [1, 2, 1, 3]
print(my_list.count(1))
````

````python
my_list = [1, 2, 3]
print(my_list.index(2))
````

````python
original = [1, 2]
new_list = original.copy()
print(new_list)
````

````python
first = [1, 2, 4, 5]
second = first
second.append(78)
first
````

## List Comprehension

````python
squares = [x*x for x in range(5)]
print(squares)
````

````python
even = [x for x in range(10) if x % 2 == 0]
print(even)
````
