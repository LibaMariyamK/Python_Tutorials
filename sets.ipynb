# Sets in Python — Practical Notebook

## Creating Sets

```python
my_set = set(('a','b','c'))
print(my_set)
type(my_set)
```

```python
fruits = {"Apple", "Banana", "Mango", "Apple"}
print(fruits)
```

```python
num = [1,2,3,4,2,3,2,3,4,3,4,5,67,9]
list(set(num))
```

```python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)
```

```python
colors = set(["Red", "Green", "Blue"])
print(colors)
```

```python
days = set(("Mon", "Tue", "Wed"))
print(days)
```

## Set Methods

```python
my_set = {1, 2, 3}
my_set.add(4)
print(my_set)
```

```python
my_set = {1, 2}
my_set.update([3, 4, 5])
print(my_set)
```

```python
my_set = {1, 2, 3}
my_set.remove(2)
print(my_set)
```

```python
my_set = {1, 2, 3}
my_set.discard(5)
print(my_set)
```

```python
my_set = {1, 2, 3}
removed = my_set.pop()
print(my_set)
print(removed)
```

```python
my_set = {1, 2, 3}
my_set.clear()
print(my_set)
```

## Set Operations

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)
```

```python
print(a & b)
```

```python
print(a - b)
```

```python
print(a ^ b)
```

## Membership & Built-in Functions

```python
fruits = {"Apple", "Banana", "Mango"}
print("Apple" in fruits)
```

```python
print(len(fruits))
```

## Subset, Superset & Disjoint

```python
x = {1, 2}
y = {1, 2, 3, 4}
print(x.issubset(y))
```

```python
print(y.issuperset(x))
```

```python
z = {10, 11}
print(x.isdisjoint(z))
```

## Converting Between Types

```python
unique_list = list({1, 2, 2, 3})
print(unique_list)
```

```python
names = ["Tom", "Jerry", "Tom", "Spike"]
unique_names = list(set(names))
print(unique_names)
```
