# Dictionaries in Python — Practical Notebook

## Creating Dictionaries

```python
my_dict = {}
print(my_dict)
type(my_dict)
```

```python
student = {"name": "Riya", "age": 21, "course": "Python"}
print(student)
```

```python
person = dict(name="Aman", age=25)
print(person)
```

```python
employee = {"name": "Kiran", "address": {"city": "Kochi", "pin": "682001"}}
print(employee)
```

## Accessing Values

```python
student = {"name": "Riya", "age": 21, "course": "Python"}
print(student["name"])
```

```python
print(student.get("grade"))
```

```python
print(student.get("grade", "Not Found"))
```

```python
print(student.keys())
```

```python
print(student.values())
```

```python
print(student.items())
```

```python
print("place" in student)
```

```python
print(len(student))
```

## Modifying Dictionaries

```python
student = {"name": "Riya", "age": 21}
student["age"] = 22
print(student)
```

```python
student = {"name": "Riya", "age": 21}
student["grade"] = "A"
print(student)
```

```python
student = {"name": "Riya", "age": 21}
student.update({"age": 23, "grade": "A"})
print(student)
```

```python
student = {"name": "Riya", "age": 21, "course": "Python"}
removed = student.pop("course")
print(student)
```

```python
student = {"name": "Riya", "age": 21}
last = student.popitem()
print(student, last)
```

```python
student = {"name": "Riya", "age": 21}
student.clear()
print(student)
```

## Combining Dictionaries

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
combined = {**dict1, **dict2}
print(combined)
```

## Converting Between Types

```python
student = {"name": "Riya", "age": 21, "course": "Python"}
print(list(student.keys()))
```

```python
print(list(student.items()))
```

```python
names = ["Riya", "Aman", "Kiran"]
marks = [88, 76, 91]
paired = dict(zip(names, marks))
print(paired)
```

## Dictionary Comprehension

```python
squares = {x: x*x for x in range(5)}
print(squares)
```
