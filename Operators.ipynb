## Assignment Operators

```python
num = 5
num += 3  # num = num + 3
print(num)
```

```python
num = 5
num -= 2   # num = num - 2
print(num)
```

```python
num = 5
num *= 6
print(num)
```

```python
num = 20
num /= 4
print(num)
```

```python
num = 20
num //== 3
print(num)
```

```python
num = 20
num %= 3
print(num)
```

```python
num = 2
num **= 3
print(num)
```

## Membership Operators

```python
fruits = ["Apple", "Banana", "Mango"]
print("Apple" in fruits)
```

```python
print("Grapes" not in fruits)
```

## Identity Operators

```python
a = [1, 2, 3]
b = a.copy()
b.append(5)
a
```

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]
print(a is b)
```

```python
print(a is c)
```

```python
print(a == c)
```

```python
a = 257
b = 257
a is b
```
