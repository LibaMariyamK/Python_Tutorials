## 1. What is a Variable?

Think of a variable as a **labelled box**. You put something inside the box (a value), and you give the box a name (the variable name) so you can find it and use it later.

For example, if you're building a billing system for a small store, you might want to remember the customer's name and the total bill amount. Instead of typing these values again and again, you store them in variables.

````
# Creating variables
customer_name = "Ravi"
total_bill = 1250.50

print(customer_name)
print(total_bill)
````

### How Python creates a variable

In Python, you don't need to "declare" a variable's type in advance (unlike some other languages). You simply assign a value using `=`, and Python figures out the type on its own. This is called **dynamic typing**.

````
# Python figures out the type automatically
age = 21          # this becomes an integer
price = 499.99     # this becomes a float
city = "Kochi"      # this becomes a string
is_open = True      # this becomes a boolean

print(age, price, city, is_open)
````

### Reassigning a variable

A variable is not permanent — you can change what's stored inside it at any time. The box stays, but you can swap what's inside.

````
shop_status = "Closed"
print("Morning:", shop_status)

shop_status = "Open"
print("Afternoon:", shop_status)
````

### Multiple assignment

Python lets you assign several variables in a single line — handy when you're setting up related values together.

````
# Assigning multiple variables at once
item_name, item_price, in_stock = "Notebook", 45, True

print(item_name)
print(item_price)
print(in_stock)
````

### Variable naming rules

Before we go further, here are the rules Python enforces for naming variables:

| Rule | Example (Valid) | Example (Invalid) |
|---|---|---|
| Must start with a letter or underscore | `_score`, `total1` | `1total` |
| Can contain letters, numbers, underscores | `student_age`, `x2` | `student-age` |
| Cannot use Python keywords | `class_name` | `class` |
| Case-sensitive | `Marks` and `marks` are different variables | — |

It's good practice to use names that describe what the variable holds — `total_bill` is clearer than `x`.

````
# This will cause an error - uncomment to see it
# 1total = 100

# This works fine
total1 = 100
print(total1)
````

## 2. Data Types

Every value in Python belongs to a **data type**. The four you'll use constantly as a beginner are:

| Type | Name | Example | Used for |
|---|---|---|---|
| `int` | Integer | `25`, `-10`, `0` | Whole numbers — age, quantity, count |
| `float` | Floating point | `3.14`, `99.5` | Numbers with decimals — price, weight, percentage |
| `str` | String | `"Hello"`, `'Kochi'` | Text — names, addresses, messages |
| `bool` | Boolean | `True`, `False` | Yes/no, on/off type answers |

There are more advanced types too (like `list`, `tuple`, `dict`) — we'll cover those in a later class. For today, we focus on these four building blocks.

````
# Examples of each data type
marks = 87                 # int
average = 78.4              # float
student_name = "Ananya"     # str
passed = True                # bool

print(marks, average, student_name, passed)
````

### A quick note on strings

Text in Python must be wrapped in quotes — either single `'...'` or double `"..."`. Both work the same way, just be consistent.

````
greeting1 = 'Good Morning'
greeting2 = "Good Morning"

print(greeting1 == greeting2)   # True - they're identical values
````

## 3. Type Checking — `type()`

Sometimes you'll want to confirm what type a value actually is, especially before performing an operation on it. Python gives you a built-in function for this: `type()`.

**Scenario:** You're processing a form where a customer entered their pin code. You want to check whether it was stored as text or as a number before using it in a calculation.

````
pin_code = "682001"

print(type(pin_code))
print(type(25))
print(type(25.5))
print(type(True))
````

You can also use `type()` inside a condition to make decisions based on a value's type.

````
value = 45.0

if type(value) == float:
    print("This is a decimal number")
else:
    print("This is not a decimal number")
````

## 4. Type Conversion (Type Casting)

Type conversion means changing a value from one data type to another. This comes up constantly in real programs.

**Scenario:** A customer enters their age on a form. Whatever they type is received by Python as text (a string) — even if they typed numbers. If you want to use that age in a calculation (like checking if they're eligible for a senior citizen discount), you first need to convert it into a number.

The common conversion functions are:

| Function | Converts to |
|---|---|
| `int()` | Integer |
| `float()` | Float |
| `str()` | String |
| `bool()` | Boolean |

````
age_text = "45"          # this is a string, not a number
print(type(age_text))

age_number = int(age_text)   # now converted to an integer
print(type(age_number))

# Now we can do maths with it
years_to_60 = 60 - age_number
print("Years left to reach 60:", years_to_60)
````

````
# Converting int to float and float to int
score = 88
score_as_float = float(score)
print(score_as_float)

price = 199.99
price_as_int = int(price)     # this simply drops the decimal part, it doesn't round
print(price_as_int)
````

````
# Converting numbers to strings - useful when joining text and numbers
quantity = 5
message = "You ordered " + str(quantity) + " items"
print(message)
````

### Careful — not every conversion works

If the text doesn't actually represent a valid number, Python will raise an error when you try to convert it. Try running the cell below and see what happens.

````
# This will raise an error - uncomment to see it
# invalid = int("hello")

# This works because the text is a valid number
valid = int("100")
print(valid)
````

## 5. Taking Input and Displaying Output

### `input()` — getting information from the user

The `input()` function pauses your program and waits for the user to type something and press Enter. Whatever they type is handed back to your program **as a string**, no matter what they typed.

This is why type conversion (Section 4) matters so much in real programs — right after taking input, you often need to convert it.

````
name = input("What is your name? ")
print("Hello,", name)
````

````
# input() always returns a string - even if the user types a number
age = input("Enter your age: ")
print(type(age))       # this will show <class 'str'>

age = int(age)          # now converted to an integer
print(type(age))        # this will show <class 'int'>
print("Next year you will be", age + 1)
````

### `print()` — displaying information

You've already been using `print()` throughout this notebook. Here are a few extra things it can do:

````
# print() can take multiple values, separated by commas
print("Name:", "Ravi", "Age:", 21)

# 'sep' changes what goes between the values
print("Kochi", "Kerala", "India", sep=" - ")

# 'end' changes what comes after the last value (default is a new line)
print("Loading", end="...")
print("Done")
````

### f-strings — the modern way to combine text and variables

Instead of joining text and variables with `+` (which requires converting numbers to strings first), you can use an **f-string**. Just put `f` before the quotes, and place variables inside `{}`.

````
product = "Notebook"
price = 45
quantity = 3
total = price * quantity

print(f"You bought {quantity} x {product} for a total of Rs. {total}")
````

## 6. A Few Other Important Things

### Comments

Comments are notes in your code that Python ignores when running the program. They're for humans — to explain *why* something is being done, not just *what*.

````
# This is a single-line comment
age = 21   # you can also add a comment after code on the same line

"""
This is a multi-line comment (a string Python doesn't assign anywhere),
often used for longer explanations
"""

print(age)
````

### Basic operators (a first look)

You'll cover these in more depth later, but here's a quick preview of arithmetic operators, since you'll already be using them for practice.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `5 + 3` → `8` |
| `-` | Subtraction | `5 - 3` → `2` |
| `*` | Multiplication | `5 * 3` → `15` |
| `/` | Division (always gives a float) | `5 / 2` → `2.5` |
| `//` | Floor division (drops the decimal) | `5 // 2` → `2` |
| `%` | Modulus (remainder) | `5 % 2` → `1` |
| `**` | Exponent (power) | `5 ** 2` → `25` |

````
a = 17
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
````

### `len()` — checking the length of text

A small but frequently used function — it tells you how many characters are in a string.

````
username = "student123"
print(len(username))
````
